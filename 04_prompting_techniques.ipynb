### 1.1 Zero-Shot Prompting

```python
# Example prompts for digital twin
zero_shot_examples = [
    {
        'task': 'activity_prediction',
        'prompt': 'Predict the next activity: I just arrived at work, and now...',
        'expected': 'work/take some minutes off'
    },
    {
        'task': 'preference_extraction',
        'prompt': 'Extract preferences from:I do prepare work schedule of the day',
        'expected': {'likes': ['confuguration of routers', 'reports'], 'dislikes': ['delays']}
    }
]

print('Zero-Shot Examples:')
for ex in zero_shot_examples:
    print(f"\nTask: {ex['task']}")
    print(f"Prompt: {ex['prompt']}")
    print(f"Expected: {ex['expected']}")
```
